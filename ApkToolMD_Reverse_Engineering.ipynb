```python
# Download and install Ghidra 12.1.4 (working version with analyzeHeadless)
import os, subprocess, urllib.request

ghidra_url = "https://github.com/NationalSecurityAgency/ghidra/releases/download/Ghidra_12.1.4_build/ghidra_12.1.4_PUBLIC_20260921.zip"
ghidra_zip = "ghidra_12.1.4_PUBLIC.zip"

if not os.path.exists("ghidra"):
    print("Downloading Ghidra 12.1.4...")
    urllib.request.urlretrieve(ghidra_url, "ghidra.zip")
    print("Extracting...")
    import zipfile
    with zipfile.ZipFile("ghidra.zip", 'r') as z:
        z.extractall("ghidra")
    # Find extracted directory
    for d in os.listdir("."):
        if d.startswith("ghidra_"):
            os.rename(d, "ghidra")
            break
    print("Ghidra installed")
else:
    print("Ghidra already installed")
```

```python
# Create DecompileMatchinc.py script for headless decompilation
decompile_script = """from ghidra.app.decompiler import DecompInterface
from ghidra.program.model.listing import FunctionManager
from ghidra.util.task import ConsoleTaskMonitor
import os

program = currentProgram
if program is None:
    print("No program loaded")
else:
    decompiler = DecompInterface()
    decompiler.openProgram(program)
    monitor = ConsoleTaskMonitor()
    
    # Decompile all functions
    fm = program.getFunctionManager()
    functions = fm.getFunctions(True)
    
    output = []
    for func in functions:
        try:
            result = decompiler.decompileFunction(func, 30, ConsoleTaskMonitor())
            if result.decompileCompleted():
                code = result.getDecompiledFunction().getC()
                output.append(f"// Function: {func.getName()} at {func.getEntryPoint()}\n{code}\n")
        except Exception as e:
            output.append(f"// Function: {func.getName()} ERROR: {e}\n")
    
    with open("/content/decompiled.c", "w") as f:
        f.write("\n".join(output))
    print(f"Decompiled {len(output)} functions to /content/decompiled.c")
"""

with open("/content/DecompileMatchinc.py", "w") as f:
    f.write(decompile_script)

print("DecompileMatchinc.py created")
```

```python
# Run Ghidra headless decompilation
import subprocess, os

ghidra_dir = "ghidra"  # extracted directory
ghidra_path = None
for d in os.listdir("ghidra"):
    if os.path.isdir(os.path.join("ghidra", d)):
        ghidra_path = os.path.join("ghidra", d)
        break

if not ghidra_path:
    print("Ghidra directory not found!")
else:
    print(f"Using Ghidra at: {ghidra_path}")

# Run headless decompilation
cmd = [
    f"{ghidra_path}/support/analyzeHeadless",
    "/content/ghidra_proj",
    "MyProject",
    "-import", "/content/libcocos2dcpp.so",
    "-processor", "AARCH64:LE:64:v8A",
    "-scriptPath", "/content",
    "-postScript", "DecompileMatchinc.py",
    "-deleteProject"
]

print("Running Ghidra headless decompilation (this may take 10-30 minutes)...")
result = subprocess.run(cmd, capture_output=True, text=True, timeout=3600)
print(result.stdout)
if result.stderr:
    print("STDERR:", result.stderr)
```

```python
# Search for pickup-related code in decompiled output
import re

if os.path.exists("/content/decompiled.c"):
    with open("/content/decompiled.c", "r") as f:
        content = f.read()
    
    # Search terms
    terms = [
        "pickup", "Pickup", "PICKUP",
        "collect", "Collect", "COLLECT",
        "item", "Item", "ITEM",
        "coin", "Coin", "COIN",
        "gem", "Gem", "GEM",
        "reward", "Reward", "REWARD",
        "onCollision", "onContact", "onTrigger",
        "onTouch", "onOverlap",
        "PhysicsContact", "PhysicsBody",
        "beginContact", "endContact",
        "setContactListener", "contactListener",
        "tag", "TAG",
        "categoryBitmask", "contactTestBitmask", "collisionBitmask",
        "removeFromParent", "removeFromWorld",
        "scheduleUpdate", "update",
    ]

matches = []
for term in terms:
    for match in re.finditer(term, content, re.IGNORECASE):
        start = max(0, match.start() - 200)
        end = min(len(content), match.end() + 200)
        context = content[start:end]
        matches.append((term, match.start(), context))

print(f"Found {len(matches)} matches")
for term, pos, ctx in matches[:20]:
    print(f"\n=== {term} at {pos} ===")
    print(ctx)

# Save matches
with open("/content/pickup_matches.txt", "w") as f:
    for term, pos, ctx in matches:
        f.write(f"=== {term} at {pos} ===\n{ctx}\n\n")
print("\nSaved to /content/pickup_matches.txt")
```

```python
# Generate Frida hook script based on findings
frida_script = """
// Frida hook for libcocos2dcpp.so - Pickup detection
// Usage: frida -U -f com.package.name -l pickup_hook.js

// Configuration - UPDATE THESE ADDRESSES FROM GHIDRA ANALYSIS
var config = {
    // Example addresses - REPLACE WITH ACTUAL ADDRESSES FROM GHIDRA
    // Find these by searching for the function names in Ghidra
    functions: {
        // Example: onContactBegin / onContactEnd
        // 'onContactBegin': ptr('0x00123456'),
        // 'onContactEnd': ptr('0x00123456'),
        // 'onContactPreSolve': ptr('0x00123456'),
        // 'onContactPostSolve': ptr('0x00123456'),
        
        // Or PhysicsContact methods
        // 'PhysicsContact::beginContact': ptr('0x00123456'),
        // 'PhysicsContact::endContact': ptr('0x00123456'),
        
        // Or Cocos2d-x PhysicsContactListener
        // 'PhysicsContactListener::onContactBegin': ptr('0x00123456'),
        // 'PhysicsContactListener::onContactEnd': ptr('0x00123456'),
        
        // Or custom pickup/item logic
        // 'Item::collect': ptr('0x00123456'),
        // 'Player::pickupItem': ptr('0x00123456'),
        // 'Item::onPickup': ptr('0x00123456'),
        // 'PickupManager::collect': ptr('0x00123456'),
    },
};

function hookFunction(name, address) {
    try {
        Interceptor.attach(address, {
            onEnter: function(args) {
                console.log(`[+] ${name} called`);
                console.log(`    args[0] = ${args[0]}`);
                console.log(`    args[1] = ${args[1]}`);
                console.log(`    this = ${this}`);
                
                // Print backtrace
                console.log(Thread.backtrace(this.context, Backtracer.ACCURATE)
                    .map(DebugSymbol.fromAddress).join('\n'));
            },
            onLeave: function(retval) {
                console.log(`[+] ${name} returned: ${retval}`);
            }
        });
        console.log(`[+] Hooked ${name} at ${address}`);
    } catch (e) {
        console.log(`[-] Failed to hook ${name}: ${e}`);
    }
}

function scanForPickup() {
    // Scan memory for strings containing pickup/collect
    var results = Memory.scanSync(/pickup|collect|collectItem|pickupItem|collectItem|onPickup/i);
    for (var i = 0; i < results.length; i++) {
        console.log(`[+] Found string at ${results[i].address}`);
        try {
            var str = results[i].address.readUtf8String();
            console.log(`    String: ${str}`);
        } catch (e) {}
    }
}

function main() {
    console.log("[+] Pickup hook loaded");
    
    // Hook known functions (UPDATE ADDRESSES FROM GHIDRA!)
    for (var name in config.functions) {
        hookFunction(name, config.functions[name]);
    }
    
    // Scan for pickup strings in memory
    scanForPickup();
    
    console.log("[+] All hooks installed");
}

setImmediate(main);
"""

with open("/content/pickup_hook.js", "w") as f:
    f.write(frida_script)

print("Frida hook script saved to /content/pickup_hook.js")
print("\nNext steps:")
print("1. Open /content/decompiled.c and search for pickup/collect/item functions")
print("2. Note the function addresses from Ghidra")
print("3. Update addresses in /content/pickup_hook.js")
print("4. Run: frida -U -f com.package.name -l /content/pickup_hook.js")
```
