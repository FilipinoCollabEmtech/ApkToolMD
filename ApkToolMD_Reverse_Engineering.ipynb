# ApkToolMD - libcocos2dcpp.so Reverse Engineering (rabin2)

Uses `rabin2 -s` to dump all symbols (function names + addresses) from `libcocos2dcpp.so`, plus `izz` for strings. Then generates a Frida hook.

**Steps:**
1. Download the binary
2. Install radare2 (rabin2)
3. `rabin2 -s` — dump all symbols
4. `rabin2 -z` — dump all strings
5. Search for pickup/aim/weapon logic
6. Generate Frida hook

```python
# Dump ALL symbols with rabin2 -s (this is what works when afl returns nothing)
import subprocess, os, re, json

binary = "libcocos2dcpp.so"

print("=== rabin2 -s (symbols) ===")
res = subprocess.run(["rabin2", "-s", binary], capture_output=True, text=True)
symbols_out = res.stdout
print(symbols_out[:5000])
print(f"\n... ({len(symbols_out)} chars total)")

print("\n=== rabin2 -z (strings) ===")
resz = subprocess.run(["rabin2", "-z", binary], capture_output=True, text=True)
strings_out = resz.stdout
print(strings_out[:5000])
print(f"\n... ({len(strings_out)} chars total)")

# Save raw outputs
with open("/content/rabin_symbols.txt", "w") as f:
    f.write(symbols_out)
with open("/content/rabin_strings.txt", "w") as f:
    f.write(strings_out)

# Parse rabin2 -s output (format: vaddr paddr len type ... name)
print("\n=== PARSED SYMBOLS (pickup/aim/weapon/item/collision) ===")
terms = [
    "pickup", "collect", "item", "coin", "gem", "reward",
    "aim", "angle", "shoot", "bullet", "weapon", "fire",
    "collision", "contact", "trigger", "touch", "overlap",
    "physics", "DidPickupWeapon", "WeaponCollision",
    "ItemLibrary", "ItemPurchase", "scheduleUpdate", "update"
]

sym_matches = []
for line in symbols_out.splitlines():
    parts = line.split()
    if len(parts) < 4:
        continue
    # rabin2 -s columns: vaddr paddr len type ... name
    vaddr = parts[0]
    name = parts[-1] if parts[-1] not in ("FUNC", "OBJ", "DATA", "NOTYPE") else ""
    if not name:
        continue
    for t in terms:
        if t.lower() in name.lower():
            sym_matches.append((t, vaddr, name))
            break

print(f"Found {len(sym_matches)} symbol matches")
for t, vaddr, name in sym_matches[:80]:
    print(f"  [{t}] {vaddr}: {name}")

# Parse rabin2 -z strings for matches
print("\n=== STRING MATCHES ===")
str_matches = []
for line in strings_out.splitlines():
    for t in terms:
        if t.lower() in line.lower():
            str_matches.append((t, line[:200]))
            break

print(f"Found {len(str_matches)} string matches")
for t, line in str_matches[:80]:
    print(f"  [{t}] {line}")

# Save parsed matches
with open("/content/rabin_matches.txt", "w") as f:
    f.write("=== SYMBOL MATCHES ===\n")
    for t, vaddr, name in sym_matches:
        f.write(f"[{t}] {vaddr}: {name}\n")
    f.write("\n=== STRING MATCHES ===\n")
    for t, line in str_matches:
        f.write(f"[{t}] {line}\n")

print("\nSaved:")
print("  /content/rabin_symbols.txt")
print("  /content/rabin_strings.txt")
print("  /content/rabin_matches.txt")
```

```python
# Generate Frida hook script
frida_script = '''// Frida hook for libcocos2dcpp.so
// Usage: frida -U -f com.package.name -l pickup_hook.js

// Configuration - FILL IN ADDRESSES FROM /content/rabin_symbols.txt
// rabin2 -s gives virtual addresses. On Android, add Module.findBaseAddress("libcocos2dcpp.so")
var config = {
    moduleName: "libcocos2dcpp.so",
    // Example format (replace with real addresses from rabin2 -s):
    // functions: {
    //   "DidPickupWeapon": ptr("0x0108ED8A"),
    //   "WeaponCollisionBegan": ptr("0x01085FA6"),
    //   "WeaponCollisionEnded": ptr("0x01085FBB"),
    // },
    functions: {},
};

function hookFunction(name, address) {
    try {
        Interceptor.attach(address, {
            onEnter: function(args) {
                console.log(`[+] ${name} called`);
                console.log(`    args[0] = ${args[0]}`);
                console.log(`    args[1] = ${args[1]}`);
                console.log(`    this = ${this}`);
                console.log(Thread.backtrace(this.context, Backtracer.ACCURATE)
                    .map(DebugSymbol.fromAddress).join('\\n'));
            },
            onLeave: function(retval) {
                console.log(`[+] ${name} returned: ${retval}`);
            }
        });
        console.log(`[+] Hooked ${name} at ${address}`);
    } catch (e) {
        console.log(`[-] Failed to hook ${name}: ${e}`);
    }
}

function main() {
    console.log("[+] Hook loaded");
    var base = Module.findBaseAddress(config.moduleName);
    console.log(`[+] ${config.moduleName} base: ${base}`);
    
    for (var name in config.functions) {
        var addr = config.functions[name];
        // If address is a string like "0x0108ED8A", treat as offset from base
        if (typeof addr === "string") {
            addr = base.add(parseInt(addr, 16));
        }
        hookFunction(name, addr);
    }
    console.log("[+] All hooks installed");
}

setImmediate(main);
'''

with open("/content/pickup_hook.js", "w") as f:
    f.write(frida_script)

print("Frida hook script saved to /content/pickup_hook.js")
print("\nNext steps:")
print("1. Open /content/rabin_symbols.txt")
print("2. Find functions you want (pickup, aim, weapon, collision, etc.)")
print("3. Copy their addresses into config.functions in /content/pickup_hook.js")
print("4. Run: frida -U -f com.package.name -l /content/pickup_hook.js")
```
