```python
# Dump EVERY function and string from libcocos2dcpp.so with radare2
import subprocess, os, json, re

binary = "libcocos2dcpp.so"

def r2_cmd(cmd):
    result = subprocess.run(
        ["r2", "-q", "-c", cmd, binary],
        capture_output=True, text=True
    )
    return result.stdout

# Step 1: Full analysis (find functions even in stripped binary)
print("Running radare2 full analysis (aaa)...")
aaa_out = r2_cmd("aaa")
print(aaa_out)

# Step 2: Dump all functions (text)
print("\n=== ALL FUNCTIONS (afl) ===")
afl = r2_cmd("afl")
print(afl)

# Step 3: Dump all functions (JSON)
print("\n=== ALL FUNCTIONS (aflj) ===")
aflj = r2_cmd("aflj")
try:
    funcs = json.loads(aflj)
    print(f"Total functions: {len(funcs)}")
    for f in funcs:
        print(f"  0x{f['offset']:08x}  {f.get('name','')}  size={f.get('size',0)}")
except Exception as e:
    print(aflj[:3000])

# Step 4: Dump all strings
print("\n=== ALL STRINGS (izz) ===")
iz = r2_cmd("izz")
print(iz[:5000])

# Step 5: Search for interesting terms in strings
print("\n=== INTERESTING STRING MATCHES ===")
terms = [
    "pickup", "collect", "item", "coin", "gem", "reward",
    "aim", "angle", "shoot", "bullet", "weapon", "fire",
    "onContact", "onCollision", "onTrigger", "onTouch", "onOverlap",
    "PhysicsContact", "PhysicsBody", "beginContact", "endContact",
    "contactListener", "collisionBitmask", "contactTestBitmask",
    "removeFromParent", "scheduleUpdate", "update",
    "DidPickupWeapon", "WeaponCollision", "ItemLibrary", "ItemPurchase"
]
try:
    strs = json.loads(r2_cmd("izzj"))
except Exception:
    strs = []

matches = []
for s in strs:
    st = s.get("string", "")
    for t in terms:
        if t.lower() in st.lower():
            matches.append((t, s.get("paddr", 0), st))
            break

print(f"Found {len(matches)} matches")
for t, addr, st in matches[:100]:
    print(f"  [{t}] 0x{addr:08x}: {st[:200]}")

# Save everything to files
with open("/content/r2_all_functions.txt", "w") as f:
    f.write("=== AAA ===\n")
    f.write(aaa_out)
    f.write("\n=== AFL (functions) ===\n")
    f.write(afl)
    f.write("\n=== AFLJ (functions json) ===\n")
    f.write(aflj)
    f.write("\n=== IZZ (strings) ===\n")
    f.write(iz)
    f.write("\n=== INTERESTING MATCHES ===\n")
    for t, addr, st in matches:
        f.write(f"[{t}] 0x{addr:08x}: {st}\n")

with open("/content/r2_strings.txt", "w") as f:
    f.write(iz)

print("\nSaved:")
print("  /content/r2_all_functions.txt")
print("  /content/r2_strings.txt")
```

```python
# Generate Frida hook script template
frida_script = '''// Frida hook for libcocos2dcpp.so - Pickup / Aim / Weapon detection
// Usage: frida -U -f com.package.name -l pickup_hook.js

// Configuration - FILL IN ADDRESSES FROM /content/r2_all_functions.txt (afl output)
var config = {
    functions: {
        // Example - replace with real addresses from radare2:
        // 'DidPickupWeapon': ptr('0x010XXXXX'),
        // 'WeaponCollisionBegan': ptr('0x010XXXXX'),
        // 'WeaponCollisionEnded': ptr('0x010XXXXX'),
        // 'ItemLibrary::getItem': ptr('0x010XXXXX'),
        // 'Player::setAimAngle': ptr('0x010XXXXX'),
    },
};

function hookFunction(name, address) {
    try {
        Interceptor.attach(address, {
            onEnter: function(args) {
                console.log(`[+] ${name} called`);
                console.log(`    args[0] = ${args[0]}`);
                console.log(`    args[1] = ${args[1]}`);
                console.log(`    this = ${this}`);
                console.log(Thread.backtrace(this.context, Backtracer.ACCURATE)
                    .map(DebugSymbol.fromAddress).join('\\n'));
            },
            onLeave: function(retval) {
                console.log(`[+] ${name} returned: ${retval}`);
            }
        });
        console.log(`[+] Hooked ${name} at ${address}`);
    } catch (e) {
        console.log(`[-] Failed to hook ${name}: ${e}`);
    }
}

function scanForStrings() {
    var patterns = [/pickup/i, /collect/i, /aim/i, /angle/i, /shoot/i, /weapon/i, /bullet/i];
    patterns.forEach(function(pat) {
        var results = Memory.scanSync(pat);
        for (var i = 0; i < results.length; i++) {
            console.log(`[+] Found string at ${results[i].address}`);
            try {
                console.log(`    ${results[i].address.readUtf8String()}`);
            } catch (e) {}
        }
    });
}

function main() {
    console.log("[+] Hook loaded");
    for (var name in config.functions) {
        hookFunction(name, config.functions[name]);
    }
    scanForStrings();
    console.log("[+] All hooks installed");
}

setImmediate(main);
'''

with open("/content/pickup_hook.js", "w") as f:
    f.write(frida_script)

print("Frida hook script saved to /content/pickup_hook.js")
print("\nNext steps:")
print("1. Open /content/r2_all_functions.txt and review ALL functions")
print("2. Note addresses for pickup, aim, weapon, collision functions")
print("3. Update addresses in /content/pickup_hook.js")
print("4. Run: frida -U -f com.package.name -l /content/pickup_hook.js")
```
