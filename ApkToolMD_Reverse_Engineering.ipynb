```python
# Analyze libcocos2dcpp.so with radare2
import subprocess, os, json, re

binary = "libcocos2dcpp.so"

def r2_cmd(cmd):
    result = subprocess.run(
        ["r2", "-q", "-c", cmd, binary],
        capture_output=True, text=True
    )
    return result.stdout

# Full analysis
print("Running radare2 analysis (aa)...")
print(r2_cmd("aa"))

# Dump functions
print("\n=== Functions (afl) ===")
afl = r2_cmd("aflj")
try:
    funcs = json.loads(afl)
    print(f"Total functions: {len(funcs)}")
    for f in funcs[:80]:
        print(f"  0x{f['offset']:08x}  {f['name']}")
except Exception as e:
    print(afl[:2000])

# Dump strings
print("\n=== Strings (izz) ===")
iz = r2_cmd("izzj")
try:
    strs = json.loads(iz)
    print(f"Total strings: {len(strs)}")
    for s in strs[:100]:
        print(f"  0x{s['paddr']:08x}: {s['string'][:140]}")
except Exception as e:
    print(iz[:2000])

# Search for pickup/collect/item related strings
print("\n=== Pickup/Collect string search ===")
terms = [
    "pickup", "collect", "item", "coin", "gem", "reward",
    "onContact", "onCollision", "onTrigger", "onTouch", "onOverlap",
    "PhysicsContact", "PhysicsBody", "beginContact", "endContact",
    "contactListener", "categoryBitmask", "collisionBitmask",
    "removeFromParent", "scheduleUpdate", "update"
]
try:
    all_strings = json.loads(iz)
except Exception:
    all_strings = []

matches = []
for s in all_strings:
    st = s.get("string", "")
    for t in terms:
        if t.lower() in st.lower():
            matches.append((t, s.get("paddr", 0), st))
            break

print(f"Found {len(matches)} matches")
for t, addr, st in matches[:60]:
    print(f"  [{t}] 0x{addr:08x}: {st[:180]}")

# Save analysis results
with open("/content/r2_analysis.txt", "w") as f:
    f.write("=== RADARE2 ANALYSIS ===\n")
    f.write(r2_cmd("aa\n"))
    f.write("\n=== FUNCTIONS (afl) ===\n")
    f.write(afl)
    f.write("\n=== STRINGS (izz) ===\n")
    f.write(iz)
    f.write("\n=== PICKUP MATCHES ===\n")
    for t, addr, st in matches:
        f.write(f"[{t}] 0x{addr:08x}: {st}\n")

print("\nSaved to /content/r2_analysis.txt")
```

```python
# Generate Frida hook script based on findings
frida_script = '''// Frida hook for libcocos2dcpp.so - Pickup detection
// Usage: frida -U -f com.package.name -l pickup_hook.js

// Configuration - UPDATE THESE ADDRESSES FROM RADARE2 ANALYSIS
var config = {
    // Find function addresses from /content/r2_analysis.txt (afl output)
    // Example: look for functions containing contact/pickup/item
    functions: {
        // 'onContactBegin': ptr('0x00123456'),
        // 'onContactEnd': ptr('0x00123456'),
        // 'PhysicsContactListener::onContactBegin': ptr('0x00123456'),
        // 'Item::collect': ptr('0x00123456'),
        // 'Player::pickupItem': ptr('0x00123456'),
    },
};

function hookFunction(name, address) {
    try {
        Interceptor.attach(address, {
            onEnter: function(args) {
                console.log(`[+] ${name} called`);
                console.log(`    args[0] = ${args[0]}`);
                console.log(`    args[1] = ${args[1]}`);
                console.log(`    this = ${this}`);
                console.log(Thread.backtrace(this.context, Backtracer.ACCURATE)
                    .map(DebugSymbol.fromAddress).join('\\n'));
            },
            onLeave: function(retval) {
                console.log(`[+] ${name} returned: ${retval}`);
            }
        });
        console.log(`[+] Hooked ${name} at ${address}`);
    } catch (e) {
        console.log(`[-] Failed to hook ${name}: ${e}`);
    }
}

function scanForPickup() {
    var results = Memory.scanSync(/pickup|collect|collectItem|pickupItem|onPickup/i);
    for (var i = 0; i < results.length; i++) {
        console.log(`[+] Found string at ${results[i].address}`);
        try {
            var str = results[i].address.readUtf8String();
            console.log(`    String: ${str}`);
        } catch (e) {}
    }
}

function main() {
    console.log("[+] Pickup hook loaded");
    for (var name in config.functions) {
        hookFunction(name, config.functions[name]);
    }
    scanForPickup();
    console.log("[+] All hooks installed");
}

setImmediate(main);
'''

with open("/content/pickup_hook.js", "w") as f:
    f.write(frida_script)

print("Frida hook script saved to /content/pickup_hook.js")
print("\nNext steps:")
print("1. Open /content/r2_analysis.txt and search for pickup/collect/item functions")
print("2. Note the function addresses from radare2 (afl output)")
print("3. Update addresses in /content/pickup_hook.js")
print("4. Run: frida -U -f com.package.name -l /content/pickup_hook.js")
```
